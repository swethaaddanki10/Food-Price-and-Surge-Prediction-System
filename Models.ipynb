```python
import pandas as pd

df = pd.read_csv("south_india_commodities_2020_2025 (2).csv")
df['Arrival_Date'] = pd.to_datetime(df['Arrival_Date'])

df.head()
```

```python
df['combo'] = (
    df['State'] + "_" +
    df['District'] + "_" +
    df['Market'] + "_" +
    df['Commodity']
)
```

```python
combo_counts = df['combo'].value_counts()

valid_combos = combo_counts[combo_counts >= 500].index

df_filtered = df[df['combo'].isin(valid_combos)]

print("Valid combinations:", len(valid_combos))
```

```python
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

def evaluate_arima(data):

    data = data[['Arrival_Date', 'Modal_Price']].copy()
    data = data.sort_values('Arrival_Date')
    data = data.set_index('Arrival_Date')

    if len(data) < 300:
        return None

    split = int(len(data) * 0.8)
    train = data[:split]
    test = data[split:]

    try:
        model = ARIMA(train, order=(5,1,0))
        model_fit = model.fit()

        predictions = model_fit.forecast(steps=len(test))

        actual = test['Modal_Price'].values

        mae = mean_absolute_error(actual, predictions)
        rmse = np.sqrt(mean_squared_error(actual, predictions))

        mape = np.mean(np.abs((actual - predictions) / actual)) * 100

        return mae, rmse, mape

    except:
        return None
```

```python
results_arima = []

for i, combo in enumerate(valid_combos):

    print(f"Processing {i+1}/{len(valid_combos)}")

    subset = df_filtered[df_filtered['combo'] == combo]

    metrics = evaluate_arima(subset)

    if metrics is not None:
        mae, rmse, mape = metrics

        results_arima.append({
            "Combination": combo,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })
```

```python
results_arima_df = pd.DataFrame(results_arima)

results_arima_df.head()
```

```python
print("ARIMA Average MAPE:", results_arima_df["MAPE"].mean())
print("ARIMA Average MAE:", results_arima_df["MAE"].mean())
print("ARIMA Average RMSE:", results_arima_df["RMSE"].mean())
```

```python
df_price_stats = []

for combo in valid_combos:
    subset = df_filtered[df_filtered['combo'] == combo]

    avg_price = subset['Modal_Price'].mean()
    std_price = subset['Modal_Price'].std()

    df_price_stats.append({
        "Combination": combo,
        "Avg_Price": avg_price,
        "Std_Price": std_price
    })

price_stats_df = pd.DataFrame(df_price_stats)

merged = results_arima_df.merge(price_stats_df, on="Combination")

clean_results = merged[merged["Std_Price"] > 10]
```

```python
top5_arima = clean_results.sort_values("MAPE").head(5)

top5_arima
```

```python
import pandas as pd

df = pd.read_csv("south_india_commodities_2020_2025 (2).csv")
df['Arrival_Date'] = pd.to_datetime(df['Arrival_Date'])
```

```python
df['combo'] = (
    df['State'] + "_" +
    df['District'] + "_" +
    df['Market'] + "_" +
    df['Commodity']
)
```

```python
combo_counts = df['combo'].value_counts()

valid_combos = combo_counts[combo_counts >= 300].index

df_filtered = df[df['combo'].isin(valid_combos)]

print("Valid combinations:", len(valid_combos))
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

def evaluate_sarima(data):

    data = data[['Arrival_Date', 'Modal_Price']].copy()
    data = data.sort_values('Arrival_Date')
    data = data.set_index('Arrival_Date')

    if len(data) < 300:
        return None

    split = int(len(data) * 0.8)
    train = data[:split]
    test = data[split:]

    try:
        model = SARIMAX(
            train,
            order=(1,1,1),
            seasonal_order=(1,1,1,7)  # weekly seasonality
        )

        model_fit = model.fit(disp=False)

        predictions = model_fit.forecast(steps=len(test))

        actual = test['Modal_Price'].values

        mae = mean_absolute_error(actual, predictions)
        rmse = np.sqrt(mean_squared_error(actual, predictions))
        mape = np.mean(np.abs((actual - predictions) / actual)) * 100

        return mae, rmse, mape

    except:
        return None
```

```python
results_sarima = []

for i, combo in enumerate(valid_combos):

    print(f"Processing {i+1}/{len(valid_combos)}")

    subset = df_filtered[df_filtered['combo'] == combo]

    metrics = evaluate_sarima(subset)

    if metrics is not None:
        mae, rmse, mape = metrics

        results_sarima.append({
            "Combination": combo,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })
```

```python
results_sarima_df = pd.DataFrame(results_sarima)

results_sarima_df.head()
```

```python
print("SARIMA Average MAPE:", results_sarima_df["MAPE"].mean())
print("SARIMA Average MAE:", results_sarima_df["MAE"].mean())
print("SARIMA Average RMSE:", results_sarima_df["RMSE"].mean())
```

```python
!pip install prophet

import pandas as pd
import numpy as np
from prophet import Prophet
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
df = pd.read_csv("south_india_commodities_2020_2025 (2).csv")
df['Arrival_Date'] = pd.to_datetime(df['Arrival_Date'])
```

```python
df['combo'] = (
    df['State'] + "_" +
    df['District'] + "_" +
    df['Market'] + "_" +
    df['Commodity']
)
```

```python
combo_counts = df['combo'].value_counts()

valid_combos = combo_counts[combo_counts >= 300].index

df_filtered = df[df['combo'].isin(valid_combos)]

print("Valid combinations:", len(valid_combos))
```

```python
def evaluate_prophet(data):

    data = data[['Arrival_Date', 'Modal_Price']].copy()
    data.columns = ['ds', 'y']
    data = data.sort_values('ds')

    if len(data) < 300:
        return None

    split = int(len(data) * 0.8)
    train = data[:split]
    test = data[split:]

    model = Prophet(
        daily_seasonality=True,
        weekly_seasonality=True,
        yearly_seasonality=True
    )

    model.fit(train)

    future = model.make_future_dataframe(periods=len(test))
    forecast = model.predict(future)

    predictions = forecast['yhat'][-len(test):].values
    actual = test['y'].values

    mae = mean_absolute_error(actual, predictions)
    rmse = np.sqrt(mean_squared_error(actual, predictions))

    mape = np.mean(np.abs((actual - predictions) / actual)) * 100

    return mae, rmse, mape
```

```python
results_prophet = []

for i, combo in enumerate(valid_combos):

    print(f"Processing {i+1}/{len(valid_combos)}")

    subset = df_filtered[df_filtered['combo'] == combo]

    metrics = evaluate_prophet(subset)

    if metrics is not None:
        mae, rmse, mape = metrics

        results_prophet.append({
            "Combination": combo,
            "MAE": mae,
            "RMSE": rmse,
            "MAPE": mape
        })
```

```python
results_prophet_df = pd.DataFrame(results_prophet)

results_prophet_df.head()
```

```python
print("Prophet Average MAPE:", results_prophet_df["MAPE"].mean())
print("Prophet Average MAE:", results_prophet_df["MAE"].mean())
print("Prophet Average RMSE:", results_prophet_df["RMSE"].mean())
```

```python
df_price_stats = []

for combo in valid_combos:
    subset = df_filtered[df_filtered['combo'] == combo]

    avg_price = subset['Modal_Price'].mean()
    std_price = subset['Modal_Price'].std()

    df_price_stats.append({
        "Combination": combo,
        "Avg_Price": avg_price,
        "Std_Price": std_price
    })

price_stats_df = pd.DataFrame(df_price_stats)

merged = results_prophet_df.merge(price_stats_df, on="Combination")

clean_results = merged[merged["Std_Price"] > 10]
```
